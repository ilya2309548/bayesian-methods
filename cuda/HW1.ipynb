ДЗ №1
Задача
Дан файл с исходным кодом классического train loop на PyTorch (лежит в task1.py рядом с этим файлом).

Логически этот скрипт работает, однако с точки зрения взаимодействия с железом содержит неэффективности. Если запустить этот код на большом датасете и большой модели, он будет работать в несколько раз медленнее нормы, а потом неизбежно упадет с ошибкой OOM (Out Of Memory).

Ваша задача - это найти проблемные места в функции train(), опираясь на знания из лекции, и переписать исходный файл так, чтобы он работал эффективно. Желательно оставить краткие комментарии в коде о том, почему вы внесли те или иные изменения.

Всего в коде есть несколько уровней проблем:

уровень 1 - утечки памяти;
уровень 2 - нарушения работы асинхронного конвейера;
уровень 3 - нечестные метрики.
Сдача
Исправленный файл с исходным кодом в виде ссылки на путь в GitHub-репозитории нужно отправить мне в личку в телеграм.

Баллы
В зависимости от того, сколько проблем вы найдете и исправите:

уровень 1 - 3 балла;
уровни 1 и 2 - 4 балла;
уровни 1, 2 и 3 - 5 баллов.
Дедлайн
Сдавать и корректировать решение можно до 29 мая.
